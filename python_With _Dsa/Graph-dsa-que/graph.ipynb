```python
from collections import deque

def bfs(graph,start):
    visited=set()
    queue=deque([start])
    while queue:
        node=queue.popleft()
        if node not in visited:
            print(node)
            visited.add(node)
            for  i in graph[node]:
                queue.append(i)
    
Edges = [[0,1],[0,2],[1,3]]
graph={}
for u,v in Edges:
    if u not in graph:
        graph[u]=[]
    if v not in graph:
        graph[v]=[]
    graph[u].append(v)
    graph[v].append(u)
start=0
bfs(graph,start)
```
