```python
# column :[Age,Math,Marks,Science Marks]
data=np.array([
    [18,85,78],
    [19,92,88],
    [17,76,95],
    [20,90,85]
])
data
```

```python
data.shape
```

```python
# find the avg age of students
np.mean(data[:,0:1])
```

```python
# Extract math marks of all student

data[:,1]
```

```python
# Find the highest science marks
np.max(data[:,2])
```

```python
# Get details of the student who scored more than 90 marks
data[data[:,1]>90]
```

```python
# Increase maths marks of student + 5
data[:,1]+=5
data
```

```python
# find how many student younger thann 19
len(data[data[:,0] < 19])
```

```python
# calculate the avg marks of each subject(colun axis=0)

np.mean(data[:,1:],axis=0)
```

```python
# get data of student who scored at leat 80 in both subject
data[(data[:,1] >=80) & (data[:,2]>=80)]
```

```python
# replace all science marks <75 with 0
data[:,2][data[:,2]<75]=0
data
```
