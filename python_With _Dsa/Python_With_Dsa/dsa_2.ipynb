```python
# 3. Check if a string is a palindrome.
s="sehos"
left,right=0,len(s)-1
palindrom=True
while left<right:
    if not s[left].isalnum():
        left+=1
    elif not s[right].isalnum():
        right-=1
    elif s[left].lower()!=s[right].lower():
        palindrom=False
        break
    else:
        left+=1
        right-=1
print(palindrom)
```

```python
# 21.Find the maximum occurring element.
lst = [1, 2, 2, 3, 2,4, 4,4,4,3]

n = len(lst)
visted = [0] * n

max_fre = 0
min_fre = n

max_ele = 0
min_ele = 0

for i in range(n):
    if visted[i]:
        continue

    cnt = 1

    for j in range(i + 1, n):
        if lst[i] == lst[j]:
            visted[j] = 1
            cnt += 1

    if cnt > max_fre:
        max_ele = lst[i]
        max_fre = cnt

    if cnt < min_fre:
        min_ele = lst[i]
        min_fre = cnt

print("max_occurance_ele:", max_ele)
print("Max_frequency:", max_fre)

print("min_occurance_ele:", min_ele)
print("minimum_frequency:", min_fre)
```
