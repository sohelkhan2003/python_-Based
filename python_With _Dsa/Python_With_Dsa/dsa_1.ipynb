```python
# selectionsort mean first index to min and same process to last index
def selectionsort(arr):

    n=len(arr)
    for i in range(n):
        min=i
        for j in range(i+1,n):
            if arr[min]>arr[j]:
                min=j
        arr[i],arr[min]=arr[min],arr[i]
    return arr
arr=[10,7,5,9,15]
selectionsort(arr)
```

```python
# merge sort
def mergesort(arr):
    if len(arr)<=1:
        return arr

    mid=len(arr)//2
    left=mergesort(arr[:mid])
    right=mergesort(arr[mid:])

    return merge(left,right)


def merge(left,right):
    result=[]
    i=j=0

    while i<len(left) and j<len(right):
        if left[i]<right[j]:
            result.append(left[i])
            i+=1
        else:
            result.append(right[j])
            j+=1

    result.extend(left[i:])
    result.extend(right[j:])

    return result


arr=[6,3,9,5,2,8]
print(mergesort(arr))
```

```python
def quicksort(nums):
    if len(nums)<=1:
        return nums
    pivot=nums[-1]

    left=[i for i in nums[:-1] if i<=pivot]
    right=[i for i in nums[:-1] if i>pivot]

    return quicksort[left]+[pivot]+quicksort[right]
arr=[6,3,9,5,2,8]
print(mergesort(arr))

```

Aapko shuruat mein kuch log diye gaye hain jo "Happy" hain. Har ek iteration (round) ke baad, kuch Happy log Sad ho jate hain aur kuch Sad log Happy (ya Sad hi rehte hain). Aapko ye track karna hai ki kuch rounds ke baad kitne log Happy hain aur kitne Sad.

Rules (Transitions):

Happy logo ke liye:

70% log Sad ho jayenge.

30% log Happy hi rahenge.

Sad logo ke liye:

50% log Sad hi rahenge.

(Iska matlab bacha hua 50% log Happy ho jayenge).

```python
N=int(input())
n=int(input())

def simultenous_state(N,n):
    happy=N
    sad=0
    for i in range(n):
        new_happy=(happy*0.3)+(sad*0.5)
        new_sad=(happy*0.7)+(sad*0.5)
        sad=new_sad
        happy=new_happy
    return int(happy),int(sad)
print(simultenous_state(N,n))
```
