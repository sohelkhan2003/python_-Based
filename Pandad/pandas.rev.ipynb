```python
# 🔥 Basic Programming / Logic Questions
# Prime Number

n=1

if n<=1:
    print("not prime")
else:
    for i in range(2,n):
        if n%i==0:
            print(" not prime")
            break
    else:
        print("prime")

```

```python
# Composite Number
n=int(input("Enter a num"))
if n<=1:
    print("neither prime nor composite")
else:
    for i in range(2,n):
        if n%i==0:
            print("Composite Num ")
            break
    else:
        print("prime")
```

```python
# Sum of Digits
n=int(input("Enter a num:"))
sum=0
while n>0:
    r=n%10
    sum+=r
    n//=10
sum
```

```python
# Reverse a Number
n=int(input("Enter  a num:"))
rev=0
while n>0:
    r=n%10
    rev=rev*10+r
    n//=10
rev
```

```python
# Palindrome Number
n=int(input("Enter  a num:"))
s=n
rev=0
while n>0:
    r=n%10
    rev=rev*10+r
    n//=10
if s==rev:
    print("Palindrom")
else:
    print("Not a palindrom")
```

```python
# Palindrome String
s="sohosds"
left,right=0,len(s)-1
while left<right:
    if not s[left].isalnum():
        left+=1
    elif not s[right].isalnum():
        right-=1
    elif s[left].lower()!=s[right].lower():
        print("Not palindrom")
        break
    else:
        left+=1
        right-=1
else:

    print("Plindrom")
```

```python
# Fibonacci Series
n=int(input("Enter a num"))
if n==0:
    print('0')
elif n==1:
    print('1')
else:
    a,b=0,1
    for i in range(2,n+1):
        a,b=b,a+b
        print(b,end=" ")
```
