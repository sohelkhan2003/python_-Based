```python
import pandas as pd
data ={'value':[10,12,13,15,18,32,5,7,300,600]}
df=pd.DataFrame(data)



# Calculator IQR
Q1=df['value'].quantile(0.25)
Q2=df['value'].quantile(0.75)
IQR =Q2-Q1

# Outlier Bounds

lower_bound=Q1 - 1.5 * IQR
upper_bound=Q2 + 1.5 * IQR

# Boolean Mask Outliers

outliers = (df['value'] < lower_bound )| (df['value'] >upper_bound)

# Count and print outlier

outlier_count =outliers.sum()
print('Total outlier: ',outlier_count)
print("Outlier values:",df[outliers])
```

```python
import numpy as np

df=pd.DataFrame({'value':[10,12,13,17,19,20,35,600,750,800]})

q1=df['value'].quantile(0.25)
q2=df['value'].quantile(0.75)

IQR=q2-q1

lower_bound=q1-1.5*IQR
upper_bound=q2+1.5*IQR

outlier=df[(df['value']>lower_bound) & (df['value']<upper_bound)]
print(outlier)
```

```python
import numpy as np

df=pd.DataFrame({'value':[10,12,13,17,19,20,35,600,750,800]})

def outlier_mean(df,col):
    q1=df[col].quantile(0.25)
    q2=df[col].quantile(0.75)

    IQR =q2-q1

    lower_bound=q1-1.5*IQR
    upper_bound=q2+1.5*IQR
#  replace  outlier  with mean of non-outlier value
    mean_value=df[(df[col]>=lower_bound) & (df[col]<=upper_bound)][col].mean()
    df[col]=df[col].apply(lambda x:mean_value if x<lower_bound or x >upper_bound else x)
    return df[col]
```

```python
import pandas as pd

df=pd.DataFrame({'value':[10,20,30,40,50,60,70,80,400,900]})

q1=df['value'].quantile(0.25)
q2=df['value'].quantile(0.75)

iqr=q2-q1

lower =q1-1.5*iqr
upper =q2+1.5*iqr

mean_vlue =df[(df['value']>lower) & (df['value']<=upper)]['value'].mean()
df['value']=df['value'].apply(lambda x:mean_vlue if x<lower or x > upper else x)
print(df)
```
