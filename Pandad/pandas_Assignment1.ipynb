Test case 1: 
Input: Series([2, 5, NaN, 3, 9, NaN, 4]) 
Output: Series([2, 5, 4.0, 3, 9, 4.0, 4]) 


```python
df=pd.Series([2,5,np.nan,3,9,np.nan,4])
m=df.median()
df.fillna(m)
```

```python
df=pd.Series([2,5,np.nan,3,9,np.nan,4])
df.fillna(4,inplace=True)
print(df)
```

Q3. Map values from another Series based on a condition: 
Test case 1: 
Input: Series([10, 20, 30, 40]), mapping: {10: 'A', 30: 'B'} 
Output: Series(['A', 20, 'B', 40]) 


```python
df=pd.Series([10,20,30,40])
mapping={10:'A',30:'B'}
df.map(mapping).fillna(df)
```

```python
s=pd.Series([10,20,30,40])
maping={10:'A',30:'B'}
s.map(maping).fillna(s)
```

Q4. Calculate cumulative product of values, ignoring zeroes: 
Test case 1: 
Input: Series([1, 2, 0, 4, 5]) 
Output: Series([1, 2, 2, 8, 40]) 


```python
import pandas as pd
df=pd.Series([1,2,0,4,5])
result=df.replace(0,1).cumprod()
print(result)
```

Q5. Write a python program which Shift elements of a Series and fill shifted positions with custom values: 
Test case 1: 
Input: Series([10, 20, 30, 40]), shift: 2, fill_value: -1 
Output: Series([-1, -1, 10, 20]) 


```python
df=pd.Series([10,20,30,40])
df.shift(2).fillna(-1).astype(int)
```

```python
s=pd.Series([10,20,30,40])
result=s.shift(2).fillna(-1).astype(int)
result
```

Q8. Find positions of prime numbers in a Series and return the count: 
Test case 1: 
Input: Series([10, 7, 3, 15, 17]) 
Output: 3 


```python
import numpy as np
import pandas as pd
```

```python
s=pd.Series([10, 7, 3, 15, 17]) 

def is_prime(n):
    if n<=1:
        return False
    for i in range(2,int(np.sqrt(n))+1):
        if n%i==0:
            return False
    return True
n=int(input("Enter  a num "))
is_prime(n)
```
