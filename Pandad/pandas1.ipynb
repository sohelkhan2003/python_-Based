Pivot Table

```python
import numpy as np
import pandas as pd
```

Pivot Table

```python


# Creating the data
data = {
    'Date': pd.date_range('2023-01-01', periods=20),
    'product': ['A', 'B', 'C', 'D'] * 5,
    'Region': ['East', 'West', 'Nort', 'South', 'East', 'West', 'Nort', 'South', 'East', 'West',
               'Nort', 'South', 'East', 'West', 'Nort', 'South', 'East', 'West', 'Nort', 'South'],
    'Sales': np.random.randint(100, 1000, 20),
    'units': np.random.randint(10, 100, 20),
    'Rep': ['john', 'Merry', 'Bob', 'Alice', 'john', 'Merry', 'Bob', 'Alice', 'john', 'Merry',
            'Bob', 'Alice', 'john', 'Merry', 'Bob', 'Alice', 'john', 'Merry', 'Bob', 'Alice']
}

# Creating the DataFrame
df = pd.DataFrame(data)

# Adding Month and Quarter columns
df['Month'] = df['Date'].dt.month_name()
df['Quarter'] = 'Q' + df['Date'].dt.quarter.astype(str)

# Display the result
df
```

```python
# aggregration are use in pivot table
pd.pivot_table(df,values='Sales',index='Region',columns='product',aggfunc='median')
```

```python
pd.pivot_table(df,values=['Sales','units'],index='Region',columns='product')
```

Cross Tabs

```python
# cross tab are use at a counting purpose
pd.crosstab(df['Region'],df['product'])
```
