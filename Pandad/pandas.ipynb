```python
df=pd.DataFrame({'value':[10,12,13,14,11,400,14,13,10,9,600,500,28,5,65]})

Q1=df['value'].quantile(0.25)
Q3=df['value'].quantile(0.75)

iqr=Q3-Q1

lower=Q1-1.5*iqr
upper=Q3+1.5*iqr
mean_value=df[(df['value']>=lower) & (df['value']<=upper)]['value'].mean()
df['value']=df['value'].apply(lambda x: mean_value if  x < lower or x > upper else x)

```
