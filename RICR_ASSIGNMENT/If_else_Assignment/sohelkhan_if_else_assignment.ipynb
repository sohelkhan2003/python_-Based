```python
n1=int(input("Enter 1 num :"))
n2=int (input ("Enter 2 num :"))
n3=int(input("Enter 3 num :"))
n4=int(input("Enter 4 num :"))

total=n1+n2+n3+n4
if total==360:
    print("It is quadrital ")
else:
    print("It is not a quadrital")
```
