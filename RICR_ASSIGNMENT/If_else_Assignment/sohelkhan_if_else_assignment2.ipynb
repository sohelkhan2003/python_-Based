Q19. Write a program to input electricity unit charges and calculate 
total electricity bill 
according to the given condition:
• For first 50 units Rs. 0.50/unit
• For next 150 units Rs. 0.75/unit
• For next 250 units Rs. 1.20/unit
• For unit above 450 Rs. 1.50/unit
• An additional surcharge of 20% is added to the bill
Test Case 1:
Input: 100 units
Expected Output:
Rs. 62.5
With Surcharge: Rs. 75
Test Case 3:
Input: 300 units
Expected Output:
Rs. 257.5
With Surcharge: Rs. 309

```python
u=float(input("Enter a unit"))

if u>0 and u<=50 :
    total_bill=(u*0.50)

elif u>50 and u<=200:
    total_bill=(50*0.50+(u-50)*0.75)

elif u>200 and u<=450:
    total_bill=(50*0.50+150*0.75+(u-200)*1.20)

elif u>450:
    total_bill=(50*0.50+150*0.75+250*1.20+(u-450)*1.50)

print(total_bill+(total_bill*0.20))
```
