2. Write a Python program that prompts the user to input a number
and handles a KeyboardInterrupt exception if the user cancels the
input.


```python
def x():
    try: 
        n=int(input("Enter a num"))
        return n
    except ValueError:
            raise Exception(" input  be integer")
    
try:
    s = x()
    
except KeyboardInterrupt:
    print("Cancelled by user")
else:
    print(s)
finally:
    print("End code")
```

7. Write a Python program that checks if a number is prime. Raise an
exception if the input is not a positive integer, and always confirm
that the check is complete

```python
def prime(n):
    if n<=1:
        return False
    for i in range(2,n):
        if n%i==0:
            return False
    return True
try:
    n=int(input("Enter a num "))

    if n>1:
        print("the num is positive")
    if n<0:
        print("The num is negative ")

    if prime(n):
        print("Prime")
    else:
        raise Exception("Not a prime")
except Exception as e:
    print(e)
finally:
    print("code end")
```
