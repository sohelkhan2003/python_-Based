sort an array of 0s,1s,2s (Dutch nATIONAL FLAG PROBLEM)

```python
lst=[0,1,0,1,0,2,1,2]
low,mid=0,0
high=len(lst)-1
while mid<=high:
    if lst[mid]==0:
        lst[low],lst[mid]=lst[mid],lst[low]
        low +=1
        mid+=1
    elif lst[mid]==1:
        mid+=1
    else:
        lst[mid],lst[high]=lst[high],lst[mid]
        high-=1
print(lst)
        
```

Que .Minimum num of jumps to reach the end

```python
def min_num_jumps(lst):

    n=len(lst)
    if n<=0:
        return 0
    if lst[0]==0:
        return -1  

    jump=0
    curend=0
    farthest=0

    for i in range(n-1):
        farthest=max(farthest,i+lst[i]) 

        if i==curend:
            jump+=1
            curend =farthest
            if curend >= n-1:
                return jump
        if farthest <= i and i<n-1:
            return -1
    return-1
lst=[2,3,1,1,4]
min_num_jumps(lst)

```

```python
def rearangeg_altermnte_ele(lst):

    pos=[i for i in lst if i >=0]
    neg=[i for i in lst if i <0]

    s=[]
    x=y=0

    while x< len(pos) and y <len(neg):
        s.append(neg[y])
        y+=1
        s.append(pos[x])

    while y <len(neg):
        s.append(neg[y])
        y+=1
    while x< len(pos):
        s.append(pos[x])
        x+=1
    return s
lst=[1,2,3,-4,-5,-1]
rearangeg_altermnte_ele(lst)
```
