Q1. Write a python program to divide the list into two equal halves
and print the sum of each half.
lst1 = [2,4,6,8,3,5,8,9]

```python
lst1 = [2,4,6,8,3,5,8,9]
mid=len(lst1)//2
s1=sum(lst1[:mid])
s2=sum(lst1[mid:])
print(s1,s2)
```

```python
lst1 = [2,4,6,8,3,5,8,9,1,2]

s1=0
s2=0
mid=len(lst1)//2
for i in range(mid):
    s1+=lst1[i]

for i in range(mid,len(lst1)):
    s2+=lst1[i]
print(s1,s2)
```

```python
lst1 = [2,4,6,8,3,5,8,9]
s1=0
s2=0
for i in range(len(lst1)):
    r=lst1[i]//2
    if i<r:
        s1+=lst1[i]
    else :
        s2+=lst1[i]
print(s1,s2)
```
