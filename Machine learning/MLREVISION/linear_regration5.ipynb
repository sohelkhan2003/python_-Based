```python
plt.scatter(df['Built-up Area (sq.ft.)'],df['Construction Cost (₹ Lakhs)'])
plt.plot(df['Built-up Area (sq.ft.)'],df['Construction Cost (₹ Lakhs)'],color='red')
plt.xlabel("Built-up Area (sq.ft.)")
plt.ylabel("Construction Cost (₹ Lakhs)")
plt.show()
```

```python
x_mean=df['Built-up Area (sq.ft.)'].mean()
y_mean=df['Construction Cost (₹ Lakhs)'].mean()
x_mean
```

```python
y_mean
```

```python
numerator=0
denomenator=0
for i,j in zip(df['Built-up Area (sq.ft.)'],df['Construction Cost (₹ Lakhs)']):
    numerator+=(i-x_mean)*(j-y_mean)
    denomenator+=(i-x_mean)**2
    
```

```python
numerator
```

```python
denomenator
```

```python
m=(numerator/denomenator)
m
```

```python
c=y_mean-m*x_mean
c
```

```python
x_new = 2000
pred_cost = m*x_new + c
pred_cost
```

```python
# 6.	1000 sq.ft. 
y_1000=m*1000+c
y_1000
```

```python
df['y_pred']=[(m*i+c) for i in df['Built-up Area (sq.ft.)']]
df['y_pred']
```

```python
df
```

```python
plt.scatter(df['Built-up Area (sq.ft.)'],df['Construction Cost (₹ Lakhs)'])
plt.plot(df['Built-up Area (sq.ft.)'],df['y_pred'],color='red')
plt.xlabel('Built-up Area (sq.ft.)')
plt.ylabel('Construction Cost (₹ Lakhs)')
plt.show()
```
