Gradisent algo

```python
def gradisent_algo(x,y):
    import matplotlib.pyplot as plt
    import numpy as np
    m=b=0
    learning_rate=0.0001
    n=len(x)
    for i in range(1,50):
        y_pred=m * x+b
        error =y-y_pred
        cost=(1/n) * sum([val**2 for val in (y - y_pred)])
        md = -(2/n) * sum((y-y_pred)*x)
        bd = -(2/n)*sum(y-y_pred)
        m=m-learning_rate*md
        b=b-learning_rate*bd
        print("m{},b {}, cost {} ,iteration {}".format(m,b,cost,i))
        plt.figure(figsize=(4,4))
        plt.scatter(x,y)
        plt.plot(x,y_pred,color='red')
        plt.show()
x=np.array([10,13,18,15,22])
y=np.array([14,17,29,21,33])

gradisent_algo(x,y)
```
