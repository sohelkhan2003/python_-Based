```python
x_mean=df['Study Hours (per day)'].mean()
y_mean=df['Marks (%)'].mean()
```

```python
x_mean
```

```python
y_mean
```

```python
numerator=0
denominator=0
for i,j in zip(df['Study Hours (per day)'],df['Marks (%)']):
    numerator+=(i-x_mean)*(j-y_mean)
    denominator+=(i-x_mean)**2
```

```python
numerator
```

```python
denominator
```

```python
m=(numerator/denominator)
m
```

```python
c=y_mean-m*x_mean
c
```

```python
# 4.	Predict marks for a student studying 6.5 hours per day. 
hour=6.5
predicted_marks=m*hour+c
predicted_marks
```

```python
df['y_pred']=[(m*i+c) for i in df['Study Hours (per day)']]
df['y_pred']
```
