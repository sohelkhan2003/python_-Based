```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.neighbors import KNeighborsRegressor
from sklearn.model_selection import train_test_split,GridSearchCV
from sklearn.preprocessing import StandardScaler,LabelEncoder

import warnings
warnings.filterwarnings('ignore')
```

```python
df=sns.load_dataset("tips")
```

```python
for i in df.select_dtypes(include='category'):
    print(df[i].unique())
```

```python
gender_encoder=LabelEncoder()
smoker_encoder=LabelEncoder()
day_encoder=LabelEncoder()
time_encoder=LabelEncoder()

df['sex']=gender_encoder.fit_transform(df['sex'])
df['smoker']=smoker_encoder.fit_transform(df['smoker'])
df['day']=day_encoder.fit_transform(df['day'])
df['time']=time_encoder.fit_transform(df['time'])
```

```python
x=df[['total_bill','sex', 'smoker', 'day', 'time', 'size']]
y=df['tip']
```

```python
x
```

```python
y
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
scaler=StandardScaler()
x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.fit_transform(x_test)
```

```python
from sklearn.metrics import r2_score
training_acc=[]
testing_acc=[]
for i in range(1,20,2):
    Knn=KNeighborsRegressor(n_neighbors=i,metric='minkowski',p=2)
    Knn.fit(x_train_scaled,y_train)
    y_train_pred=Knn.predict(x_train_scaled)
    y_pred=Knn.predict(x_test_scaled)
    training_acc.append(r2_score(y_train,y_train_pred))
    testing_acc.append(r2_score(y_test,y_pred))
```

```python
for i , j in zip(training_acc,testing_acc):
    print(i,j)

    
```
