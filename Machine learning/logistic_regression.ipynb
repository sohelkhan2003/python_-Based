```python
# irq
q1=df['TotalCharges'].quantile(0.25)
q3=df['TotalCharges'].quantile(0.75)
iqr=q3-q1
lower=q1-1.5*iqr
upper=q3+1.5*iqr

mean_vaue=df[(df['TotalCharges']>lower)&(df['TotalCharges']<upper)]['TotalCharges'].mean()
print(mean_vaue)
df.loc[:,'TotalCharges'] =df['TotalCharges'].apply(lambda x:mean_vaue if x<lower or x>upper else x)
```

```python
cutoff_thr=pd.DataFrame(columns=['prob','accuracy','sensi','speci','precision', 'f1_score'])
from sklearn.metrics import confusion_matrix
import numpy as np
from sklearn.metrics import precision_score,recall_score,accuracy_score,f1_score
import matplotlib.pylab as plt

thresholds =np.arange(0.0,1.0,0.01)
scores=[]

for thresh in thresholds:
    y_pred =(pred_prob_y >= thresh).astype(int)
    precision=precision_score(y_test, y_pred, zero_division=0)
    recall = recall_score(y_test, y_pred)
    accuracy = accuracy_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred, zero_division=0)

    # Calculate specificity
    cm = confusion_matrix(y_test, y_pred)
    tn, fp, fn, tp = cm.ravel()
    specificity = tn / (tn + fp)

    scores.append((thresh, accuracy, recall, specificity, precision, f1))

cutoff_thr = pd.DataFrame(scores, columns=['prob', 'accuracy', 'sensi', 'speci', 'precision', 'f1_score'])

print(scores)
```
