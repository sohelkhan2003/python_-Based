```python
mean_x=df['House Size (sq.ft.)'].mean()
mean_y=df['Price (₹ Lakhs)'].mean()
```

```python
numerator
```

```python
denominator
```

```python
m=(numerator/denominator)
m
```

```python
c=mean_y-m*mean_x
c
```

```python
# the price of 1250 sq.ft
y_1250=m*1250+c
# # the price of 1250 sq.ft.house
y_1750=m*1750+c 
print("the price of 1250 sq.f.house t",y_1250/100000,"(₹ Lakhs)")
print("the price of 1750 sq.ft.house",y_1750,"(₹ Lakhs)")
```
