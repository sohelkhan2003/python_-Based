```python
df['Price (₹ Lakhs)']=df['Price (₹ Lakhs)']*10000
```

```python
mean_x=df['House Size (sq.ft.)'].mean()
mean_y=df['Price (₹ Lakhs)'].mean()
```

```python
numerator=0
denomitor=0
for i,j in zip(df['House Size (sq.ft.)'],df['Price (₹ Lakhs)']):
    numerator+=(i-mean_x)*(j-mean_y)
    denomitor+=(i-mean_x)**2
```

```python
numerator
```

```python
denomitor
```

```python
m=(numerator/denomitor)
```

```python
c=mean_y-m*mean_x
c
```

```python
y_1250=m*1250+c
y_1750=m*1750+c
print(y_1250/10000)
print(y_1750/10000)
```
