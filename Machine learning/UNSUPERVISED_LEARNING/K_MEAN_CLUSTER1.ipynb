```python
import seaborn as sns
from sklearn.cluster import KMeans
```

```python
df=sns.load_dataset("iris")
df
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.drop("species",inplace=True,axis=1)
```

```python
km=KMeans()
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score as sk_silhouette_score
```

```python
km=KMeans(n_clusters=5)
```

```python
km.fit(df)
```

```python
wcss=[]
silhouette=[]
for k in range(2,5):
    km=KMeans(n_clusters=k)
    km.fit(df)
    wcss.append(km.inertia_)
    label=km.predict(df)
    silhouette.append(sk_silhouette_score(df,labels=label))

    
```

```python
silhouette
```

```python
wcss
```

```python
import matplotlib.pyplot as plt
plt.plot(range(2,5),wcss)
```

```python
plt.plot(range(2,5),silhouette)
plt.legend()
plt.show()
```
