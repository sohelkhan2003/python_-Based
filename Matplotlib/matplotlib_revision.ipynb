CUSTOMIZATION:

1 Bar color:change using color:'green'
2. bar Width : Changing using Width =0.8.
3. Align: Position of x axis content using align='edge' or 'center'
4.

```python
plt.bar(x,y,color='green',width=0.8,align='edge')
```

```python
z=[75,85,65,34]
p=np.arange(len(x))

width =0.2
p1=[i+ width for i in p]

plt.bar(p,y,color='orange',width=0.2)
plt.bar(p1,z,color='red',width=0.2)


plt.title("programing languare  vs popularty")
plt.xlabel("language")
plt.ylabel("popularty")
plt.xticks(p+width/2, x)
plt.show()
```

HORIZONTAL  BAR GRAPH 

```python
plt.barh(x,y,color='orange',label='popularty')


plt.title("programing languare  vs popularty")
plt.xlabel("language")
plt.ylabel("popularty")
plt.legend()
plt.show()
```

```python
x=[1,2,3,4]
y=[10,20,30,40]
z=[7,8,9,5]
```

```python
fig,ax = plt.subplots(nrows=2,ncols =2)

ax[0,0].plot(x,y)
ax[1,0].plot(x,y)
ax[1,1].plot(x,z)
ax[1,1].plot(x,z)
```

```python
fig,ax=plt.subplots(1,2,figsize=(10,5))

x=[1,2,3,4]
y=[10,20,15,25]

ax[0].plot(x,y)
ax[0].set_title('Line plot')

ax[1].bar(x,y)
ax[1].set_title('Bar chart')

plt.tight_layout()
plt.show()
```
