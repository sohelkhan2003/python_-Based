Matplotlib  Basic

```python
x=np.linspace(0,5,11)
x
```

```python
y= x**3
y
```

```python
plt.plot(x,y)
plt.title("Hello ")
plt.xlabel("Chacha")
plt.ylabel("Chachi")
plt.show()
```

subplots in matplotlib

```python
plt.subplot(2,2,1)
plt.plot(x,y)
plt.title("Sohel")
plt.subplot(2,2,4)
plt.plot(y,x)
plt.ylabel("Ayan")
plt.subplot(2,2,3)
plt.plot(x**3,y**3)
plt.title("Naziya")
plt.subplot(2,2,2)
plt.plot(x**2,y**2)
plt.title("Saniya")
```

Object oriented way

Matplotlib allow the aspect ration, DPI and ig size to be specified whe the figure is created .YOu can use the fig size ans dpi keyword argument.

Figsize is a tuple of the width and height of the figure in inches
dpi is the dots-per-inch (pixel per inch) 

```python
fig= plt.figure(figsize=(12,8),dpi=100)
# left ki vLUE and bottom ki value and 
axis1=fig.add_axes([0.1,0.1,0.8,0.8])
axis1.plot(x,y)
axis2=fig.add_axes([0.3,0.3,.3,.3])
axis2.plot(x,y)
```

```python
fig=plt.figure()
axis1=fig.add_axes([0.2,0.2,0.8,0.8])
axis1.plot(x,y)
axis2=fig.add_axes([0.4,0.5,.3,.3])
axis2.plot(x,y)
```

```python
fig=plt.figure()
axis1=fig.add_axes([0.2,0.2,1,1])
axis1.plot(x,y)
axis2=fig.add_axes([0.4,0.5,.3,.3])
axis2.plot(x,y)
```

```python
xaxis= np.linspace(0,5,11)
xaxis
```

```python
fig,ax=plt.subplots(figsize=(12,6))
ax.plot(xaxis,xaxis+1,color='red',linewidth=0.25)
ax.plot(xaxis,xaxis+2,color='red',linewidth=0.50)
ax.plot(xaxis,xaxis+3,color='red',linewidth=1.00)
ax.plot(xaxis,xaxis+4,color='red',linewidth=2.00)

# possible linestype option '-',
ax.plot(xaxis,xaxis+5,color='green',lw=3,linestyle='-')
ax.plot(xaxis,xaxis+6,color='green',lw=3,linestyle='-.')
ax.plot(xaxis,xaxis+7,color='green',lw=3,ls=':')

# Custom dash
line, =ax.plot(xaxis,xaxis+8,color='black',lw=1.50)
line.set_dashes([5,50,15,50])# format :line length ,scape lengtg ...........

# possible marker symbols: marker ="+","o","*","/"

ax.plot(xaxis,xaxis+9,color='blue',lw=3,ls='-',marker="+")
ax.plot(xaxis,xaxis+10,color='blue',lw=3,ls='--',marker="o")
ax.plot(xaxis,xaxis+11,color='blue',lw=3,ls='-',marker="s")
ax.plot(xaxis,xaxis+12,color='blue',lw=3,ls='--',marker="1")
ax.plot(xaxis,xaxis+13,color='purple',lw=3,ls='-',marker="o",markersize=2)
ax.plot(xaxis,xaxis+14,color='purple',lw=3,ls='-',marker="1",markersize=4)
ax.plot(xaxis,xaxis+15,color='blue',lw=3,ls='-',marker="o",markersize=8,markerfacecolor='red')
ax.plot(xaxis,xaxis+16,color='purple',lw=1,ls='-',marker="s",markersize=8,markerfacecolor='yellow',
        markeredgecolor='green')
```
