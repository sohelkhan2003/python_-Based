Scatter plot 

```python
x=[1,2,3,4,5,6]
y=[4,5,6,1,8,9]

plt.title('Scatter plot')
plt.xlabel('Month')
plt.ylabel('Number')

C=['red','magenta','orange','blue','green','brown']
plt.scatter(x,y,color=C)
plt.show()
```

```python
x=[1,2,3,4,5,6]
y=[4,5,6,1,8,9]

plt.title('Scatter plot',color='r')
plt.xlabel('Month',color='g')
plt.ylabel('Number',color='g')

C=['red','magenta','orange','blue','green','brown']
plt.scatter(x,y,color=C,s=150)
plt.show()
```

```python
x=[1,2,3,4,5,6]
y=[4,5,6,1,8,9]

plt.title('Scatter plot',color='r')
plt.xlabel('Month',color='g')
plt.ylabel('Number',color='g')

# C=['red','magenta','orange','blue','green','brown']
plt.scatter(x,y,color='g',s=50)
plt.show()
```

Matplotlib pie chart

```python
import matplotlib.pyplot as plt
```

```python
x=[10,20,30,40]
y=['english','Hindi','Maths','Reasoning']


plt.pie(x,labels=y)
plt.show()
```

```python
x=[10,20,30,40]
y=['english','Hindi','Maths','Reasoning']

c=['yellow','brown','black','aqua']
plt.pie(x,labels=y,colors=c)
plt.show()
```
