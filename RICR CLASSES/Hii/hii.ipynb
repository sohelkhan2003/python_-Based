```python
def fun(lst):
    s=[lst[-1]]
    for i in range(len(lst)-2,-1,-1):
        if lst[i]>lst[i+1]:
            s.append(lst[i])
    s.reverse()
    return s
lst=[35,4,30,7,17,25,24,23,10,5]
fun(lst)
```
