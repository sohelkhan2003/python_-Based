HIGHEER ORDER FUNCTION

```python
def cube(x):
    return x**3
def squ(x):
    return x**2
def transform(fun,squ):
    lst1=[]
    for i in squ:
        lst1.append(fun(i))
    return lst1
lst=[1,2,3,4,5]
print(transform(squ,lst))
print(transform(cube,lst))
```

```python
# use lamda function
def transform(fun,seq):
    lst1=[]
    for i in seq:
        lst1.append(fun(i))
    return lst1
lst=[1,2,3,4,5]
print(f'list of square -> {transform(lambda x:x**2,lst)}')
print(f'list of cube->{transform(lambda x:x**3,lst)}')
```

```python
def divisible(lst):
    five=0
    three=0
    seven=0
    for i in lst:
        if i%3==0:
            three+=1
        if i%5==0:
            five+=1
        if i%7==0:
            seven+=1
    return (three,five,seven)
lst=[3,5,7,12,15,36,21,56,65]
print(divisible(lst))
```

```python
def divisible(fun,lst):
    sum=0
    for i in lst:
        if fun(i):
            sum+=1
    return sum
lst=[3,5,7,12,15,36,21,56,65]
x=lambda x:x%3==0
y=lambda y:y%5==0
z=lambda z:z%7==0
print(divisible(x,lst))
print(divisible(y,lst))
print(divisible(z,lst))
```

```python
def function(fun,lst):
    sum=0
    for i in lst:
        if fun(i):
            sum+=1
    return sum
lst=[3,5,7,12,15,36,21,56,65]
x=lambda x:x%3==0
y=lambda y:y%5==0
z=lambda z:z%7==0
print(function(x,lst))
print(function (y,lst))
print(function (z,lst))
```

reduce()

a reduce fun operatrion perform computation and sequential pair of value in sequence and return the result

```python
from functools import reduce
lst=[24,57,23,1,5]
s=reduce(lambda x,y:x+y,lst)
print(s)
```

```python
# minimum value in a list
lst=[24,57,23,1,5]
s=reduce(lambda x,y:x if x<y else y,lst)
print(s)
```
