```python
def sum_of_digit(n):
    sum=0
    while n>0:
        r=n%10
        sum+=r
        n//=10
    return sum

n=int(input("Enter a digit"))   
while True:
    if n<=9:
        print(n)
        break
    n=sum_of_digit(n)
```
