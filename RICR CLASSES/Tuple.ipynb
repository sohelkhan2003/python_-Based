```python
# Empty tuple
t=()
print(t)

# tuple having integer
t =(1,2,3)
print(t)

# tuple with mixed data type
t=(1,2,3,"sohel",2.6,74)
print(t)
```

TUPLE HAS NO ATTRIBUTE HAS A SORT

TUPLE ARE A VALID MIN AND MAX AND SUM

```python
t=(1,2,3,4,6,7,7)

for i in range(1,len(t)):
    s=t[i-1]
    if s>t[i]:
        print("not sorted")
        break
else:
    print("sorted")

        
       
```
