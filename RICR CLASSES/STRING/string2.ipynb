Q1. Write a Python program to check if a given string is a valid email
address.
Ex: example@domain.com
Hint: Check ‘@’ comes earlier than ‘dot’

```python
def check_validmail(s):
    if '@' in s and '.' in s:
        first=s.index('@')
        sec=s.index('.')
    if first<sec:
        return True
    else:
        return False
s=input("Enter a email")

if check_validmail(s):
    print("email is valid")
else:
    print("Email is not valid")
```

Q8. Write a Python program to capitalize the first and last character
of every word in a given string.
Input: "hello world from python"
Output: "HellO WorlD FroM PythoN"

```python
def first_last_char(s):
    word=s.split()
    new_word=[]
    for i in word:
        if len(i)==1:
            new_word.append(i.title())
        else:
            new_word.append(i[0].upper()+i[1:-1]+i[-1].title())
    return ' '.join( new_word)
s='hello world from python'
first_last_char(s)
```

```python
# title area use in a first letter area  upper case convert
def first_last_char(x):
    word=x.split()
    s=[]
    for i in word:
        if len(i)==1:
            s.append(i.title())
        else:
            s.append(i[0].upper()+i[1:-1]+i[-1].title())
    return ' '.join(s)
x='hello world from python'
first_last_char(x)
```
