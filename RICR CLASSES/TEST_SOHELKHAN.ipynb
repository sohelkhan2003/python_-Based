Q6. Write a program to check whether a number is Armstrong or not.

```python
n=int(input("Enter a num :"))
i=n
sum=0
while n>0:
    n//=10
    sum+=1
ans=0
n=i
while n>0:
    r=n%10
    ans+=r**sum
    n//=10
if ans==i:
    print("Armstrong")
else:
    print("Not Armstrong")
```
