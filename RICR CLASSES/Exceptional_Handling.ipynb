```python
# ->To handle the Exception
#  try except block 
# -> error raised in  try block will redirect to except block

try:
    lst=[12,23,14,15,16,4]
    print(lst[90])
except :
    print("Index is out of range ,Please enter valid index")
```

```python
# multiple exception
try:
    lst=[12,14,15,18,11,25]
    print(lst[90]) #Index error
    print(A) #name error
    print(5/0)# zero
except NameError:
    print("Variable is not defined")
except IndexError:
    print("Index ids out of range :please enter valid index")
except ZeroDivisionError:
    print("Cannot divide any num with zero")
```

Generic Exception handling

```python
# Generic Exception handling
# This is a generally used  
try:
    lst=[12,14,15,18,11,25]
    print(lst[90]) #Index error
    print(A) #name error
    print(5/0)# zero
except NameError:
    print("Variable is not defined")

except ZeroDivisionError:
    print("Cannot divide any num with zero")
except Exception as e:
    print(e)

```

```python
try:
    lst=[12,14,15,18,11,25]
    print(lst[90]) #Index error
    print(A) #name error
    print(5/0)# zero
except Exception as e:
    print(e)
except NameError:
    print("Variable is not defined")

except ZeroDivisionError:
    print("Cannot divide any num with zero")
```

Else block -> success in try will redirect else block

```python
try:
    lst=[12,14,15,18,11,25]
    print('the element at index 3 is ',lst[3]) #no error else
    # print(lst[50])
except Exception :
    print("code fatt gaya")
else:
    print("Try successfully")
```

Finally block

```python
# ->this block will definitely execute even when error occured or not occured 
# -> this s agenerally added to end of the code so that user program can interpret that the code is execute completely
try:
    lst=[12,14,15,18,11,25]
    # print('the element at index 3 is ',lst[3]) #no error else
    print(lst[3])
except Exception :
    print("code fatt gaya")
else:
    print("Try successfully")
finally :
    print("program End")
```

Raise

-> we can raise anerror by own using raise keyword

```python
def sum(a,b):
    if b<0:
        raise Exception("positive num do na bhai ")
    elif a<0:
        raise Exception("positive num do na bhai")
    return (a+b)


try:
    ans=sum(4,5)
except Exception as e:
    print(e)
else:
    print(ans)
finally :
    print('COde end')
```
