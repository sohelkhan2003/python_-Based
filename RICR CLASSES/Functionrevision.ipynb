```python
def fact(x):
    ans=1
    for i in range(1,x+1):
        ans*=i
    return ans
n=int(input("Enter  a n num "))
r1=int(input("Enter a  r value"))
n1=fact(n)
r2=fact(r1)
nminsur=fact(n-r1)
print('ncr =',n1/r2*nminsur)
```

```python
def  cube(x):
    return x**3
def square(x):
    return x**2
def transform( fun,seq):
    lst1=[]
    for i in seq:
        lst1.append(fun(i))
    return lst1
lst=[1,2,3,4,5]

print(transform(cube,lst))
print(transform(square,lst))
print(f"List of cube :-{transform(cube,lst)}")
print(f"List of square :-{transform(square,lst)}")
```
