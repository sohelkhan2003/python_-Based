ADD ELEMENT TO A SET
WE CAN ELEMENT SINGLE ELEMENT TO add()
WE CAN ELEMENT multiple element  to use update()


```python
# method 1 add()
s={1,2}
s.add((1,5,53))

print(s)
```

```python
# method 2 update()
s={12,5,7}
s.update((15,2,3),[4,5,6,78],{1,4,83.5},"HI","FUHGFFUG")
print(s)
```

delete element in set 
1.remove() if value is not present show error
2. discard() no error

```python
s={1,2,3}
s.discard(4)
print(s)
```

```python
s={1,2,4,}
s.remove(2)
print(s)
```

```python
# POP () DELETE ANY RANDOM ELEMENT AND RETURN  THE DELETE VALUE ERROR
# clear() TO DEL ALL THE VALUE INSIDE THE SET 
# del () TO DELET THE SET
s={1,4,5,3}
a=s.pop()
print(a)
s.clear()
print(type(s))
print(s)
del(s)
```
