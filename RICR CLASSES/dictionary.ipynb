```python
# sum of key or value
s={2:4,3:9,4:16,5:25}
ansv=0
ansk=0
for i in s.keys():
    ansk+=i
    ansv+=s[i]
print("ans key ",ansk,"ans value",ansv)
```

```python
s={2:4,3:9,4:16,5:25}
ansk=0
ansv=0
for i in s.items():
    ansk+=i[0]
    ansv+=i[1]
print(ansk,ansv)
```

Q3. Write a Python program to create a dictionary using dictionary
comprehension where the keys are numbers from 1 to 10, and the
values are "even" or "odd" depending on the key value.

```python
x=[1,2,3,4,5,6,7,8,9]
s={i:('even' if i%2==0 else 'odd')   for i in x }
print(s)
```
