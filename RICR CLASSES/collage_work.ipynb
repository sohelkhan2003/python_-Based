```python
num1=float(input("Enter 1 num "))
operator=input("Enter a operator(+,-,*,/,%)")
num2=float(input("Enter 2 num "))

if operator=='+':
    result=num1+num2

elif operator=='-':
    result=num1-num2

elif operator=='*':
    result=num1*num2

elif operator=='/':
    result=num1/num2

elif operator=='%':
    result=num1%num2
else:
    result="Invalid  operator"

print("result =",result)
```

```python
def decimal_to_binary(decimal_num):
    return bin(decimal_num)[2:]

def binary_to_decimal(binary_str):
    return int(binary_str, 2)

def hexadecimal_to_binary(hex_str):
    return bin(int(hex_str, 16))[2:]

def binary_to_hexadecimal(binary_str):
    return hex(int(binary_str, 2))[2:].upper()

def main():
    print("Number Conversion Program")
    print("1: Decimal to Binary")
    print("2: Binary to Decimal")
    print("3: Hexadecimal to Binary")
    print("4: Binary to Hexadecimal")
    
    choice = int(input("Enter your choice (1-6): "))
    
    if choice == 1:
        decimal_num = int(input("Enter a decimal number: "))
        print("Binary:", decimal_to_binary(decimal_num))
    elif choice == 2:
        binary_str = input("Enter a binary number: ")
        print("Decimal:", binary_to_decimal(binary_str))
    
    elif choice == 3:
        hex_str = input("Enter a hexadecimal number: ")
        print("Binary:", hexadecimal_to_binary(hex_str))
    elif choice == 4:
        binary_str = input("Enter a binary number: ")
        print("Hexadecimal:", binary_to_hexadecimal(binary_str))
    else:
        print("Invalid choice. Please try again.")
    
if __name__ == "__main__":
    main()
```
