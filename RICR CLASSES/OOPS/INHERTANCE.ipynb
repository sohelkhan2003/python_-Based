Multiple Inheritance

```python
class BaseClass:
    def __init__(self,name,age,adress):
        self.name=name
        self.age=age
        self.adress=adress
        print("Base class constructor")
class DrivedClass(BaseClass):
    def __init__(self,name,age,adress,height):
        super().__init__(name,age,adress)
        self.height=height
        print("Drived Class constructor")
obj=BaseClass("Sohel",21,"Betul")
obj2=DrivedClass("sohel",21,"Bhopal",5.7)
```

```python
print(obj.name,obj.age,obj2.height)
```

```python
class vehical:
    def __init__(self,vehicalbrand,model,year):
        self.vehicalbrand=vehicalbrand
        self.model=model
        self.year=year
    def display_info(self):
        print("Vehical brand is",vehicalbrand,"is model is",model,"is year is",year)
class Car(vehical):
    def __init__(self,vehicalbrand,model,year,fuel_type,Seating_capacity):
        super().__init__(vehicalbrand,model,year)
        self.fuel_type=fuel_type
        self.seating_capacity=Seating_capacity
    
    def is_electric(self):
        return self.fuel_type.lower()=="electric"
    def display_info(self):
        super().display_info()
        print("Its fuel type is",self.fuel_type,"And seeting capacity",self.seating_capacity)
# s=vehical("toyots","t20",2024)
# s1=Car("petrol",25,)
```

Practice Question: Vector Class using Operator Overloading

Create a class called Vector that represents a 2D vector (with x and y components).
Implement operator overloading to allow vector addition using the + operator.

🔸 Requirements:
Class: Vector

Instance Variables:
x – x-component of the vector
y – y-component of the vector
Constructor:
_init_(self, x, y) → to initialize vector components
Method Overloading:
_add_(self, other) → overload + operator so that adding two vectors returns a new Vector object with added components
Display Method:
display(self) → print vector in format: "Vector(x, y)"

```python
class vector:
    def __init__(self,x,y):
        self.x=x
        self.y=y
    def __add__(self,other):
        return vector (self.x +other.x,self.y+other.y)
    def display(self):
        print(self.x,self.y)
v1=vector(5,6)
v2=vector(5,6)

v3=v1+v2

v3.display()
```

```python
class vector:
    def __init__(self,x,y):
        self.x=x
        self.y=y
    def __add__(self,other):
        return (self.x +other.x,self.y+other.y)
    def display(self):
        print(self.x,self.y)
v1=vector(5,6)
v2=vector(5,6)

v3=v1+v2
print(type(v3))
```
