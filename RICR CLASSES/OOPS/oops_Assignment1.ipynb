Que1..Define an abstract Appliance class with an abstract method operate(). Create WashingMachine and Microwave subclasses with specific operations.


```python
from abc import ABC,abstractmethod

class Appliance(ABC):

    @abstractmethod
    def operate(Abc):
        Abc.operate

class Washing_machine(Appliance):
    def operate(self):
        return 'Washing cloths ...'

class Microvave(Appliance):

    def operate(self):
        return 'Heating food ...'
    
s1=Washing_machine()
s2=Microvave()

print(s1.operate())
print(s2.operate())
```

```python
from abc import ABC,abstractmethod

class Application(ABC):

    @abstractmethod
    def operate(ABC):
        ABC.operate


class WashingMachine(Application):

    def operate(self):
        return 'Washing cloths ...'
    
class Microwave(WashingMachine):

    def operate(self):
        return 'Heating Food ...'
    
s1=WashingMachine()
s2=Microwave()

print(s1.operate())
print(s2.operate())








    
```

```python
from abc import ABC,abstractmethod
import math

class polgon(ABC):

    @abstractmethod
    def area(self):
        pass

class  Triangle(polgon):

    def __init__(self,base,height):
        self.base=base
        self.height=height
    
    def area(self):
        return 0.5 * self.base*self.height
    
class Pentagon(polgon):

    def __init__(self,side):
        self.side=side
        
    def area(self):
        return (5 * self.side ** 2) / (4 * math.tan(math.pi / 5))


s1=Triangle(10,20)
s2=Pentagon(30)

print('Triangle area',s1.area())
print('Pentage area ',s2.area())
```
