```python
class circle:
    pi=3.14
    def __init__(self,r):
        self.r=r
    def area(self):
        return circle.pi*self.r*self.r
c1=circle(5)
c2=circle(7)
print(c1.area())
print(c2.area())
```

```python
class Rectangle:
    width=0
    height=0

    def __init__(self,width,height):
        self.width=width
        self.height=height

    @classmethod
    def create_square(cls,side_length):
        return cls(side_length,side_length)

# create a recangle obj using the class constructor
r1=Rectangle(5,6)
print(r1.width,r1.height)
r2=Rectangle.create_square(5)
print(r2.height,r2.width)


```

```python
class Time:
    def __init__(self,h,m):
        self.h=h
        self.m=m
    def display_Time(self):
        print("Time Taken is ",self.h," Hour a and ",self.m,"Minutes")
    @classmethod
    def minutes(cls,min):
        h1=min//60
        m1=min%60
        return cls(h1,m1)
t1=Time(5,30)
t2=Time.minutes(150)
t2.display_Time()
t1.display_Time()

```

```python
class Time:
    def __init__(self,h,m):
        self.h=h
        self.m=m
    def display_Time(self):
        print("Time Taken is ",self.h," Hour a and ",self.m,"Minutes")
    @classmethod
    def minutes(cls,min):
        h1=min//60
        m1=min%60
        return cls(h1,m1)
    
    def hour(cls,hour):
        # s1=hour//60
        return cls(hour,0)
t1=Time(5,30)
t1.display_Time()
t2=Time.hour(120)
t2.display()
```

```python
class Student:
    def __init__(self,name,grade):
        self.name=name
        self.grade=grade
    def display(self):
        print("The name is",self.name,"and grade",self.grade)
    @classmethod
    def doc(cls,d):
        k= list(d.keys())[0]
        v=list(d.values())[0]
        return cls(k,v)
s=Student.doc({'sohel':"20"})
print(s.name)
print(s.grade)

```
