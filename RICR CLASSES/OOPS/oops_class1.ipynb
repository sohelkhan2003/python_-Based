INHERITANCE

when area class(child/drives) drives properties & method of another  class(parent/base)

Single level inheritance

```python
# single parent class
class car:
    
    color='black'

    @staticmethod
    def Start():
        print("Car started..")

    @staticmethod
    def Stop():
        print("Car is stop..")

# single child classs
class toyotacar(car):
    def __init__(self,name):
        self.name=name

car1=toyotacar('Fortuner')
car2=toyotacar("prius")

print(car1.name)
print(car1.Start())
print(car1.Stop())
print(car1.color)
```

SUPER METHOD

super() method id used to access method of  the parent.


```python
class Car:
    def __init__(self,type):
        self.type=type


    @staticmethod
    def start():
        print("Car started..")
    @staticmethod
    def Stop():
        print("Car is stop..")

class toyotacar(Car):
    def __init__(self,name,type):
        super().__init__(type)
        self.name=name
        super().start()

car1=toyotacar('Forturner','Diesel')
print(car1.name)
print(car1.type)
```

Polymorphism : Operator overloading 

When the same operator is allowed to have different meaning according to context 

operator and and dunder fun

```python
class complex:

    def __init__(self,real,imare):
        self.real=real
        self.image=imare

    def showNum(self):
        print(self.real,'i +',self.image,'j')

    def __add__(self,num2):
        newReal =self.real +num2.real
        newImg=self.image+num2.image
        return complex(newReal,newImg)
    

    def __sub__(self,num2):
        newReal =self.real -num2.real
        newImg=self.image-num2.image
        return complex(newReal,newImg)
    
    def __mul__(self,num2):
        newReal =self.real *num2.real
        newImg=self.image*num2.image
        return complex(newReal,newImg)


num1=complex(1,3)
num1.showNum()

num2=complex(4,5)
num2.showNum()

num3=num1+num2
num3.showNum()

num3=num1-num2
print(num3.showNum())

num3=num1*num2
print(num3.showNum())

# num3=num1.add(num2)
# num3.showNum()
```

Q3. Create  class called order which store item & its price.

 order1>order2 if price of order1 > price of order2

```python
class Order():
    def __init__(self,item,price):
        self.item=item
        self.price=price
    def __gt__(self,c2):
        return self.price>c2.price
    
c1=Order('Apple',100)
c2=Order('Anar',80)
print(c1<c2)

```
